# Credit card fraud EDA

Quick exploration of the kaggle creditcard.csv dataset. Goal: understand class imbalance, look at feature distributions, find anything obvious before modelling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_style('whitegrid')
```

```python
df = pd.read_csv('../data/raw/creditcard.csv')
print(df.shape)
df.head()
```

```python
# class imbalance
vc = df['Class'].value_counts()
print(vc)
print('fraud rate:', vc[1] / vc.sum())
ax = vc.plot.bar()
ax.set_title('Class distribution (0=not fraud, 1=fraud)')
```

## PCA-feature distributions for fraud vs not-fraud

Pick a few V's and compare densities.

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, feat in zip(axes.ravel(), ['V1','V2','V3','V4','V10','V14']):
    sns.kdeplot(df[df.Class==0][feat], ax=ax, label='not fraud')
    sns.kdeplot(df[df.Class==1][feat], ax=ax, label='fraud')
    ax.set_title(feat)
    ax.legend()
fig.tight_layout()
```
